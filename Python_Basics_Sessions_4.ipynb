```python
#wAp ask the user take 3 numbers and find the avg
n1=eval(input("Enter ur n1: "))
n2=eval(input("Enter ur n2: "))
n3=eval(input("Enter ur n3: "))
avg=(n1+n2+n3)/3
avg1=round(avg,2)
print(f"Avg of {n1},{n2},{n3} is:{avg1}")
```
