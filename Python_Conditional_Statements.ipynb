**if-elif-else**

```python
- when only **if** is available : one answer

    
- when **if-else** is available : one condition two possible
    

- when **if-elif-else** is available : two condition , 3 answer possible    
```

```python
if con1:
    ans1
elif con2:
    ans2
else:
    ans3
```

```python
# ask the user enter a percentage 
# if percentage greater than 75 or equal to 75 then print A
# if percentage between 60 to 74 then print  B
# if percentage between 45 to 59 then print  C
# if percentage between 35 to 44 then print  D
# otherwise print fail
per=eval(input("enter the percentage:"))
if per >= 75:
    print("A")
elif per >= 60:
    print("B")
elif per >= 45:
    print("C")
elif per >= 35:
    print("D")
else:
    print("Fail")
```

```python
# Ask the user enter age 
# if age greater thsn 100 print unlucky
# if age between 60 to 100 print ss
# if age between 30 to 60 print MA
# if age between 20 to 30 print young
# if age between 13 to 19 to print teen
# else print kid
age=eval(input("Enter the age: "))
if age > 100:
    print("unlucky")
elif age >= 60:
    print("SS")
elif age >= 30:
    print("MA")
elif age >= 20:
    print("young")
elif age >= 13:
    print("teen")
else:
    print("kid")
```
